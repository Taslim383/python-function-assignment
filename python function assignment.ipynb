```python
#Theory Questions

# 1. What is the difference between a function and a method in Python?
''' In Python, a function is a block of reusable code that performs a specific task. 
It is defined using the def keyword and is not associated with any object or class by default. 
Functions are called by their name and can take arguments and return values.
For example, a simple function like 
def greet(name):
    return "Hello, " + name 
can be called directly as greet("Alice").

On the other hand, a method is also a function, but it is specifically associated with an object and is defined inside a class.
Methods operate on the data (attributes) contained within the class and require at least one parameter—typically named self—which refers to the instance calling the method. 
we invoke a method using dot notation, such as object.method_name().
class Greeter:
    def greet(self, name):
        return f"Hello, {name}!"

g = Greeter()
print(g.greet("Alice"))
'''

# 2. Explain the concept of function arguments and parameters in Python.

''' In Python, parameters and arguments are terms used when working with functions, and although they are closely related, they have distinct meanings.

Parameters are the names listed in a function’s definition.
They act as placeholders for the values the function will receive when it is called.
For example,
def introduce(name, age):
    print(f"My name is {name} and I am {age} years old.")
introduce("Alice", 25).

Arguments, on the other hand, are the actual values passed to the function when it is invoked.
For instance, in the function call greet("Alice"), the value "Alice" is the argument passed to the name parameter.'''

#3. What are the different ways to define and call a function in Python?
'''
functions can be defined and called in various ways depending on the purpose and flexibility needed.
Here’s an explanation of the different ways to define and call a function:

1. Standard Function Definition
Use the def keyword to define a function with parameters, and call it using its name.
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")

2. Function with Default Arguments
we can assign default values to parameters so that the caller may skip them.
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet()          # Output: Hello, Guest
greet("Ravi")    # Output: Hello, Ravi

3. Function with Positional and Keyword Arguments
Functions can be called by passing arguments either by position or by explicitly naming them.
def show_info(name, age):
    print(f"{name} is {age} years old.")

show_info("Amit", 25)             # Positional
show_info(age=30, name="Seema")  # Keyword

4. Variable-Length Arguments
Use *args to accept multiple positional arguments and **kwargs to accept multiple keyword arguments.
def display(*args, **kwargs):
    print("Positional:", args)
    print("Keyword:", kwargs)
display("Pen", "Notebook", price=50, color="Blue")

5. Lambda Functions (Anonymous Functions)
For short, one-line functions, use lambda.
square = lambda x: x * x
print(square(4))  # Output: 16

6. Recursive Function
A function can call itself, useful for tasks like factorial or Fibonacci.
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))  # Output: 120

7. Method Inside a Class
Functions defined within a class are called methods and are called using object instances.
class Calculator:
    def add(self, x, y):
        return x + y

c = Calculator()
print(c.add(3, 5))  # Output: 8
'''
#4. What is the purpose of the `return` statement in a Python function?

'''
The return statement in a Python function is used to send a result back to the place where the function was called. 
It terminates the function and optionally passes back a value (or multiple values). 
Without a return statement, a function returns None by default.

Example 1: Return a Single Value

def add(a, b):
    return a + b

result = add(5, 3)
print(result)  # Output: 8

Example 2: Return Multiple Values

def calculate(a, b):
    return a + b, a * b

sum_result, product_result = calculate(4, 5)
print("Sum:", sum_result)          # Output: Sum: 9
print("Product:", product_result)  # Output: Product: 20

Example 3: Without Return

def greet(name):
    print(f"Hello, {name}!")

result = greet("Alice")
print(result)  # Output: Hello, Alice! \n None
'''

#5. What are iterators in Python and how do they differ from iterables?
'''
In Python, iterables and iterators are both used to loop through data, but they serve different roles.

An iterable is any Python object capable of returning its elements one by one. Examples include lists, tuples, strings, dictionaries, and sets. Any object that implements the __iter__() method is considered iterable. We can use a for loop to iterate over such objects directly.

An iterator, on the other hand, is an object that represents a stream of data. It keeps track of the current position during iteration and returns items one at a time using the __next__() method. Once all items are exhausted, calling next() raises a StopIteration exception. An iterator must implement both the __iter__() and __next__() methods.
Example:
numbers = [1, 2, 3]    
it = iter(numbers)     

print(next(it))  # Output: 1
print(next(it))  # Output: 2
print(next(it))  # Output: 3

Difference Between Iterable and Iterator:

  1 .Definition:

Iterable: An object that can be looped over using a for loop.

Iterator: An object that returns elements from an iterable, one at a time.

  2. Methods Implemented:

Iterable: Must implement the __iter__() method.

Iterator: Must implement both __iter__() and __next__() methods.

  3. Usage:

Iterable: Used to get an iterator using the iter() function.

Iterator: Used to get the next item using the next() function.

  4. State Management:

Iterable: Does not maintain any internal state.

Iterator: Maintains internal state to keep track of the current position.

  5. Reusability:

Iterable: Can be reused to create multiple iterators.

Iterator: Can be used only once; after exhaustion, it cannot be reset.

  6. Examples:

Iterable: Lists, tuples, dictionaries, strings, sets.

Iterator: Objects returned by calling iter() on an iterable.

  7. StopIteration Exception:

Iterable: Does not raise StopIteration.

Iterator: Raises StopIteration when there are no more items.
'''

#6. Explain the concept of generators in Python and how they are defined.

'''
A generator is a function that yields values instead of returning them.
Unlike regular functions that return a value and exit, a generator function pauses at each yield statement and resumes from there on the next call. 
This makes them more efficient for looping over large or complex data sets.

Defining a Generator Function
A generator is defined just like a normal function, but instead of using the return statement, it uses yield.

Example:

def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1
Calling this function doesn’t execute it immediately. Instead, it returns a generator object:

gen = count_up_to(3)
print(next(gen))  # Output: 1
print(next(gen))  # Output: 2
print(next(gen))  # Output: 3
print(next(gen))  # Raises StopIteration
'''
#7. What are the advantages of using generators over regular functions?

'''
    1. Memory Efficiency
Generators do not store the entire result in memory. Instead, they yield one item at a time, which makes them ideal for working with large or even infinite data streams.
Example: Instead of storing a list of 1 million numbers, a generator yields one number at a time, saving memory.

    2. Lazy Evaluation
Generators use lazy evaluation, meaning values are produced only when requested. This improves performance and makes wer program faster for large computations.

    3. Simplified Code with yield
Generators allow we to write clean, readable code for iterators without the complexity of managing __iter__() and __next__() methods manually.

A few lines using yield can replace an entire iterator class.

    4. Infinite Sequence Support
Generators are perfect for generating infinite sequences (like Fibonacci numbers or counters) because they don’t try to load everything into memory.

    5. Automatic State Management
Each time yield is called, the generator remembers its execution state (local variables, instruction pointer), making it easy to pause and resume execution.
    
    6. Faster Execution for Iteration
For large loops or pipelines of data, generators often execute faster because of reduced memory usage and on-demand evaluation.    
'''
#8. What is a lambda function in Python and when is it typically used?

'''
A lambda function in Python is a small, anonymous function defined using the lambda keyword. Unlike regular functions created using def, a lambda function can have only one expression and does not have a name (unless assigned to a variable). It is mainly used for short, simple operations where a full function definition is unnecessary.

Syntax: lambda arguments: expression
Example:
add = lambda x, y: x + y
print(add(2, 3))  # Output: 5
 
When is Lambda Typically Used?
Lambda functions are usually used in situations where a small, throwaway function is required for a short time, such as:

    1. With built-in functions like map(), filter(), and sorted()

nums = [1, 2, 3, 4, 5]
squares = list(map(lambda x: x**2, nums))
print(squares)  # Output: [1, 4, 9, 16, 25]
    
    2. As a key function in sorting

data = [(1, 'apple'), (3, 'banana'), (2, 'cherry')]
sorted_data = sorted(data, key=lambda x: x[0])
print(sorted_data)
    3. With filter() to filter elements

evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)  # Output: [2, 4]

    4. In GUI programming or event handling (e.g., Tkinter callbacks)
'''

#9. Explain the purpose and usage of the `map()` function in Python.
'''
The map() function in Python is used to apply a function to every item of an iterable (like a list or tuple) and return a new iterator with the results.
It allows for efficient data processing without the need for an explicit loop.

Purpose of map()
The main purpose of map() is to transform each element of an iterable by applying a specified function to it. 
This is especially useful when we want to perform the same operation on every element.
 Syntax: map(function, iterable)

Example 1: Using map() with a built-in function
numbers = [1, 2, 3, 4]
squares = map(lambda x: x**2, numbers)
print(list(squares))  # Output: [1, 4, 9, 16]

Example 2: Using map() with multiple iterables
a = [1, 2, 3]
b = [4, 5, 6]
result = map(lambda x, y: x + y, a, b)
print(list(result))  # Output: [5, 7, 9]

Example 3: Using a regular function with map()
def to_upper(word):
    return word.upper()

words = ['hello', 'world']
upper_words = map(to_upper, words)
print(list(upper_words))  # Output: ['HELLO', 'WORLD']
'''
#10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

'''
In Python, map(), filter(), and reduce() are built-in functions used for processing iterables, each serving a unique purpose.
The map() function applies a given function to each item of an iterable and returns a new iterator with the transformed items.
For example, if we have a list numbers = [1, 2, 3, 4] and we want to square each number, we can use map() like this: squares = map(lambda x: x**2, numbers), which will give we [1, 4, 9, 16] when converted to a list.

The filter() function, on the other hand, is used to filter items in an iterable based on a condition.
It applies a function that returns either True or False, and only the elements that return True are included in the output. 
For instance, to filter out only even numbers from the same list, we can write: evens = filter(lambda x: x % 2 == 0, numbers), resulting in [2, 4].

Finally, reduce() is used to apply a function cumulatively to the items of an iterable, reducing the iterable to a single value.
It is part of the functools module and is commonly used for operations like summing or multiplying all elements.
For example, using reduce(lambda x, y: x + y, numbers) will return 10, which is the sum of all elements in the list.
'''

#
```

```python
#practical questions

#1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list

def sum_even_numbers(numbers):
    return sum(num for num in numbers if num % 2 == 0)

numbers = [47, 11, 42, 13, 8, 30]
result = sum_even_numbers(numbers)
print("Sum of even numbers:", result)
'''output: Sum of even numbers: 80'''

#2. Create a Python function that accepts a string and returns the reverse of that string.

def reverse_string(s):
    return s[::-1]

result = reverse_string("hello")
print(result)  # Output: "olleh"


#3. Implement a Python function that takes a list of integers and returns a new list containing the squares of each number.
def square_numbers(numbers):
    return [num ** 2 for num in numbers]

numbers = [1, 2, 3, 4]
result = square_numbers(numbers)
print(result)  # Output: [1, 4, 9, 16]


# 4. Write a Python function that checks if a given number is prime or not from 1 to 200.
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True
    
for i in range(1, 201):
    if is_prime(i):
        print(i, end=" ")  # Outputs primes between 1 and 200


#5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.
class FibonacciIterator:
    def __init__(self, n):
        self.n = n
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < self.n:
            result = self.a
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return result
        else:
            raise StopIteration

fibonacci = FibonacciIterator(10)
for num in fibonacci:
    print(num, end=" ")  # Output: Fibonacci sequence up to 10 terms

# 6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

def powers_of_two(exponent):
    n = 0
    while n <= exponent:
        yield 2 ** n
        n += 1

for power in powers_of_two(5):
    print(power, end=" ")  # Output: 1, 2, 4, 8, 16, 32

#7. Implement a generator function that reads a file line by line and yields each line as a string.
def read_file_line_by_line(filename):
    with open(filename, 'r') as file:
        for line in file:
            yield line.strip()

# Assuming 'example.txt' is a file with content.
for line in read_file_line_by_line("example.txt"):
    print(line)


#8. Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.

data = [(1, 2), (3, 1), (4, 4)]
sorted_data = sorted(data, key=lambda x: x[1])

# Output: [(3, 1), (1, 2), (4, 4)]
print(sorted_data)


#9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.
def celsius_to_fahrenheit(celsius_list):
    return list(map(lambda c: (c * 9/5) + 32, celsius_list))

celsius = [0, 10, 20, 30]
fahrenheit = celsius_to_fahrenheit(celsius)
print(fahrenheit)  # Output: [32.0, 50.0, 68.0, 86.0]


#10. Create a Python program that uses `filter()` to remove all the vowels from a given string.

def remove_vowels(s):
    vowels = "aeiouAEIOU"
    return ''.join(filter(lambda x: x not in vowels, s))

# Example usage
input_string = "hello world"
result = remove_vowels(input_string)
print(result)  # Output: "hll wrld"

#Write a Python program using lambda and map.

    # List of numbers
numbers = [1, 2, 3, 4, 5]

# Use lambda and map to square each number
squared_numbers = list(map(lambda x: x ** 2, numbers))

# Print the result
print("Original List:", numbers)
print("Squared List:", squared_numbers)
```
